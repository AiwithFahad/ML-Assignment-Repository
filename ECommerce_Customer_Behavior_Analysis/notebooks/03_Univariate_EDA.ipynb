# 03 - Univariate Analysis

Examining each feature individually using histograms, KDE plots, box/violin plots (numerical) and count/pie charts (categorical).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
sns.set_palette('viridis')
plt.rcParams['figure.figsize'] = (10, 5)

df = pd.read_csv('../dataset/cleaned_ecommerce_data.csv')
df.head()
```

## Numerical Features
### Age Distribution

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
sns.histplot(df['age'], kde=True, bins=20, ax=axes[0,0], color='steelblue')
axes[0,0].set_title('Age - Histogram + KDE', fontweight='bold')
axes[0,0].set_xlabel('Age'); axes[0,0].set_ylabel('Count')

sns.kdeplot(df['age'], fill=True, ax=axes[0,1], color='darkorange')
axes[0,1].set_title('Age - KDE Plot', fontweight='bold')
axes[0,1].set_xlabel('Age')

sns.boxplot(x=df['age'], ax=axes[1,0], color='lightgreen')
axes[1,0].set_title('Age - Box Plot', fontweight='bold')

sns.violinplot(x=df['age'], ax=axes[1,1], color='salmon')
axes[1,1].set_title('Age - Violin Plot', fontweight='bold')

plt.tight_layout()
plt.savefig('../screenshots/03_age_distribution.png', dpi=100)
plt.show()
print(df['age'].describe())
```

**Observation:** Customer age is fairly evenly spread between roughly 18 and 70, with no extreme skew, indicating the platform serves a broad adult demographic rather than a narrow age niche.

### Total Spend (Order Value) Distribution

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
sns.histplot(df['total_spend'], kde=True, bins=30, ax=axes[0,0], color='steelblue')
axes[0,0].set_title('Total Spend - Histogram + KDE', fontweight='bold')

sns.kdeplot(df['total_spend'], fill=True, ax=axes[0,1], color='darkorange')
axes[0,1].set_title('Total Spend - KDE Plot', fontweight='bold')

sns.boxplot(x=df['total_spend'], ax=axes[1,0], color='lightgreen')
axes[1,0].set_title('Total Spend - Box Plot', fontweight='bold')

sns.violinplot(x=df['total_spend'], ax=axes[1,1], color='salmon')
axes[1,1].set_title('Total Spend - Violin Plot', fontweight='bold')

plt.tight_layout()
plt.savefig('../screenshots/03_spend_distribution.png', dpi=100)
plt.show()
print(df['total_spend'].describe())
```

**Observation:** Order values range broadly, with a fairly uniform spread and a long-ish right side, suggesting a mix of low-ticket and high-ticket purchases rather than one dominant price tier.

### Delivery Time (Days) Distribution

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
sns.histplot(df['delivery_days'], kde=True, bins=14, ax=axes[0], color='steelblue')
axes[0].set_title('Delivery Days - Histogram', fontweight='bold')

sns.boxplot(x=df['delivery_days'], ax=axes[1], color='lightgreen')
axes[1].set_title('Delivery Days - Box Plot', fontweight='bold')

sns.violinplot(x=df['delivery_days'], ax=axes[2], color='salmon')
axes[2].set_title('Delivery Days - Violin Plot', fontweight='bold')

plt.tight_layout()
plt.savefig('../screenshots/03_delivery_distribution.png', dpi=100)
plt.show()
print(df['delivery_days'].describe())
```

### Customer Rating Distribution

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
sns.histplot(df['rating'], kde=True, bins=20, ax=axes[0], color='steelblue')
axes[0].set_title('Rating - Histogram', fontweight='bold')

sns.boxplot(x=df['rating'], ax=axes[1], color='lightgreen')
axes[1].set_title('Rating - Box Plot', fontweight='bold')

sns.violinplot(x=df['rating'], ax=axes[2], color='salmon')
axes[2].set_title('Rating - Violin Plot', fontweight='bold')

plt.tight_layout()
plt.savefig('../screenshots/03_rating_distribution.png', dpi=100)
plt.show()
print(df['rating'].describe())
```

**Observation:** Ratings span the full 1-5 scale fairly evenly, with no strong clustering around any one score — satisfaction is mixed across the order base rather than universally positive or negative.

## Categorical Features
### Gender

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
sns.countplot(data=df, x='gender', ax=axes[0], order=df['gender'].value_counts().index, palette='viridis')
axes[0].set_title('Gender - Count Plot', fontweight='bold')
for c in axes[0].containers:
    axes[0].bar_label(c)

df['gender'].value_counts().plot.pie(autopct='%1.1f%%', ax=axes[1], colors=sns.color_palette('viridis', 3))
axes[1].set_title('Gender - Pie Chart', fontweight='bold')
axes[1].set_ylabel('')

plt.tight_layout()
plt.savefig('../screenshots/03_gender_distribution.png', dpi=100)
plt.show()
print(df['gender'].value_counts())
```

**Q: Which gender is most common?** The counts above show the relative share of orders attributable to each gender group; the most frequent category is printed in the cell output.

### Product Category (proxy for 'Membership Type' segment view)

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
order_cat = df['category'].value_counts().index
sns.countplot(data=df, y='category', ax=axes[0], order=order_cat, palette='magma')
axes[0].set_title('Product Category - Count Plot', fontweight='bold')

df['category'].value_counts().plot.pie(autopct='%1.1f%%', ax=axes[1], colors=sns.color_palette('magma', 6))
axes[1].set_title('Product Category - Pie Chart', fontweight='bold')
axes[1].set_ylabel('')

plt.tight_layout()
plt.savefig('../screenshots/03_category_distribution.png', dpi=100)
plt.show()
print(df['category'].value_counts())
```

### Payment Method

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.countplot(data=df, x='payment_method', ax=axes[0], order=df['payment_method'].value_counts().index, palette='crest')
axes[0].set_title('Payment Method - Count Plot', fontweight='bold')
axes[0].tick_params(axis='x', rotation=20)

df['payment_method'].value_counts().plot.pie(autopct='%1.1f%%', ax=axes[1], colors=sns.color_palette('crest', 5))
axes[1].set_title('Payment Method - Pie Chart', fontweight='bold')
axes[1].set_ylabel('')

plt.tight_layout()
plt.savefig('../screenshots/03_payment_distribution.png', dpi=100)
plt.show()
print(df['payment_method'].value_counts())
```

### Satisfaction Level (derived) & Returned Orders

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.countplot(data=df, x='satisfaction_level', ax=axes[0], order=['Low','Medium','High'], palette='coolwarm')
axes[0].set_title('Satisfaction Level - Count Plot', fontweight='bold')
for c in axes[0].containers:
    axes[0].bar_label(c)

df['is_returned'].map({0:'Not Returned', 1:'Returned'}).value_counts().plot.pie(
    autopct='%1.1f%%', ax=axes[1], colors=['#4CAF50', '#E53935'])
axes[1].set_title('Returned Orders - Pie Chart', fontweight='bold')
axes[1].set_ylabel('')

plt.tight_layout()
plt.savefig('../screenshots/03_satisfaction_returns.png', dpi=100)
plt.show()
print(df['satisfaction_level'].value_counts())
print(df['is_returned'].value_counts())
```

## Univariate Analysis Summary

- **Age & spend** are broadly distributed with no single dominant segment, indicating a diverse customer base.
- **Delivery time** clusters within a realistic short-to-medium window with a few longer-tail cases worth monitoring for logistics performance.
- **Ratings** are mixed across the 1-5 scale, meaning satisfaction is not guaranteed by default and warrants the bivariate investigation in Notebook 04.
- **Categorical splits** (gender, category, payment method) show the relative popularity of each segment, which will be cross-referenced against spend and satisfaction next.
