## Initial Observations

- The dataset contains **5,000 order-level records** from an e-commerce platform.
- Each row represents a single **order/transaction**, capturing the purchasing customer's demographics (age, gender), the product category, payment method, order value, delivery time, customer satisfaction rating, and whether the order was returned.
- There is **no explicit `customer_id`** column — analysis is therefore performed at the **order level**, and "customer segments" refer to groups of orders sharing demographic/behavioral attributes (e.g., all orders by Female customers, all orders in the Electronics category).
- Columns observed: `order_id`, `customer_age`, `customer_gender`, `product_category`, `payment_method`, `order_value_usd`, `delivery_time_days`, `customer_rating`, `returned`, `order_date`.
